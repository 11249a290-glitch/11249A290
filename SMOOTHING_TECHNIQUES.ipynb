```python
corpus = [
    "the quick brown fox jumps over the lazy dog",
    "the lazy dog sleeps all day long",
    "a quick brown fox runs fast",
    "the cat jumps over the fence",                            #11249A290
    "a lazy cat sleeps on the mat"
]
tokenized_corpus = [sentence.lower().split() for sentence in corpus]
all_words = [word for sentence in tokenized_corpus for word in sentence]
vocabulary = set(all_words)
V = len(vocabulary)  # Unique vocabulary size
N = len(all_words)   # Total number of words
unigram_counts = {}
bigram_counts = {}
for word in all_words:
    unigram_counts[word] = unigram_counts.get(word, 0) + 1
for sentence in tokenized_corpus:
    for i in range(len(sentence) - 1):
        bigram = (sentence[i], sentence[i+1])
        bigram_counts[bigram] = bigram_counts.get(bigram, 0) + 1
print(f"--- Corpus Info ---")
print(f"Vocabulary Size (V): {V}")
print(f"Total Words (N): {N}\n")
def calculate_smoothed_prob(word1, word2, k):
    count_w1_w2 = bigram_counts.get((word1, word2), 0)
    count_w1 = unigram_counts.get(word1, 0)
    probability = (count_w1_w2 + k) / (count_w1 + (k * V))
    return probability
test_cases = [("lazy", "dog"), ("quick", "cat")]
print("--- Add-1 (Laplace) Smoothing Results (k = 1) ---")
for w1, w2 in test_cases:
    prob = calculate_smoothed_prob(w1, w2, k=1.0)
    print(f"P('{w2}' | '{w1}') = {prob:.4f}")
print("\n--- Add-k (Lidstone) Smoothing Results (k = 0.1) ---")
for w1, w2 in test_cases:
    prob = calculate_smoothed_prob(w1, w2, k=0.1)
    print(f"P('{w2}' | '{w1}') = {prob:.4f}")
```

```python
corpus = [
    "the quick brown fox jumps over the lazy dog",
    "the lazy dog sleeps all day long",
    "a quick brown fox runs fast",
    "the cat jumps over the fence",
    "a lazy cat sleeps on the mat"                     #11249A290
]
tokenized_corpus = [sentence.lower().split() for sentence in corpus]
all_words = [word for sentence in tokenized_corpus for word in sentence]
vocabulary = set(all_words)
V = len(vocabulary)  # Unique vocabulary size
N = len(all_words)   # Total number of words
unigram_counts = {}
bigram_counts = {}
for word in all_words:
    unigram_counts[word] = unigram_counts.get(word, 0) + 1
for sentence in tokenized_corpus:
    for i in range(len(sentence) - 1):
        bigram = (sentence[i], sentence[i+1])
        bigram_counts[bigram] = bigram_counts.get(bigram, 0) + 1
print(f"--- Corpus Info ---")
print(f"Vocabulary Size (V): {V}")
print(f"Total Words (N): {N}\n")
def calculate_interpolated_prob(word1, word2, l1=0.7, l2=0.3):
    count_w1_w2 = bigram_counts.get((word1, word2), 0)
    count_w1 = unigram_counts.get(word1, 0)
    p_bigram = count_w1_w2 / count_w1 if count_w1 > 0 else 0
    count_w2 = unigram_counts.get(word2, 0)
    p_unigram = count_w2 / N
    return (l1 * p_bigram) + (l2 * p_unigram)
def calculate_stupid_backoff_prob(word1, word2, alpha=0.4):
    count_w1_w2 = bigram_counts.get((word1, word2), 0)
    count_w1 = unigram_counts.get(word1, 0)
    if count_w1 > 0 and count_w1_w2 > 0:
        return count_w1_w2 / count_w1
    count_w2 = unigram_counts.get(word2, 0)
    return alpha * (count_w2 / N)
def calculate_normal_backoff_prob(word1, word2, D=0.75):
    count_w1 = unigram_counts.get(word1, 0)
    count_w2 = unigram_counts.get(word2, 0)
    if count_w1 > 0:
        count_w1_w2 = bigram_counts.get((word1, word2), 0)
        unique_words_after_w1 = sum(1 for (b1, b2) in bigram_counts if b1 == word1)
        alpha_w1 = (D * unique_words_after_w1) / count_w1
        if count_w1_w2 > 0:
            return (count_w1_w2 - D) / count_w1
        else:
            return alpha_w1 * (count_w2 / N)
    else:
        return count_w2 / N
test_cases = [("lazy", "dog"), ("quick", "cat")]
print("--- Bigram Linear Interpolation Results (l1=0.7, l2=0.3) ---")
for w1, w2 in test_cases:
    prob = calculate_interpolated_prob(w1, w2, l1=0.7, l2=0.3)
    print(f"P('{w2}' | '{w1}') = {prob:.4f}")
print("\n--- Bigram Stupid Backoff Results (alpha = 0.4) ---")
for w1, w2 in test_cases:
    score = calculate_stupid_backoff_prob(w1, w2, alpha=0.4)
    print(f"Score('{w2}' | '{w1}') = {score:.4f}")
print("\n--- Bigram Normal Backoff Results (D = 0.75) ---")
for w1, w2 in test_cases:
    prob = calculate_normal_backoff_prob(w1, w2, D=0.75)
    print(f"P('{w2}' | '{w1}') = {prob:.4f}")
```
