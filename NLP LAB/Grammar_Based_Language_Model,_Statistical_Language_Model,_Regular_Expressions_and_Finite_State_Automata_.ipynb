```python
import re
resume = """
John David
Software Engineer
Email: johndavid@gmail.com
Phone: +91 9876543210
Skills:
Python, Java, SQL, HTML, CSS, JavaScript, Git, Pandas, NumPy
"""
name = resume.strip().split("\n")[0]
email = re.search(r'[\w\.-]+@[\w\.-]+\.\w+', resume)
email = email.group() if email else "Not Found"
phone = re.search(r'(\+91[- ]?)?[6-9]\d{9}', resume)
phone = phone.group() if phone else "Not Found"
skills_list = [
    "Python", "Java", "C", "C++", "SQL",
    "HTML", "CSS", "JavaScript",
    "Git", "GitHub", "NumPy",
    "Pandas", "Excel", "Machine Learning"
]
skills = [skill for skill in skills_list if skill.lower() in resume.lower()]
print("===== Resume Information =====")
print("Name :", name)
print("Email :", email)
print("Phone :", phone)
print("Skills :", ", ".join(skills))
```

```python
def dfa(string):
    state = "q0"
    for ch in string:
        if state == "q0":
            if ch == 'a':
                state = "q1"
            else:
                return False
        elif state == "q1":
            if ch == 'b':
                state = "q1"
            else:
                return False
    return state == "q1"
tests = ["a", "ab", "abb", "abbb"]
for s in tests:
    print(s, "Accepted" if dfa(s) else "Rejected")
```

```python
def dfa_simulator():
    string = input("Enter a string: ")
    state = "q0"
    for ch in string:
        if state == "q0":
            if ch == 'a':
                state = "q1"
            else:
                state = "Dead"
        elif state == "q1":
            if ch == 'b':
                state = "q1"
            else:
                state = "Dead"
        elif state == "Dead":
            break
    if state == "q1":
        print("Accepted")
    else:
        print("Rejected")
dfa_simulator()
```
