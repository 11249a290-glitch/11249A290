```python
# Program 1: Word Classes and Basic PoS Tagging using NLTK (Fixed)
import nltk
from nltk import word_tokenize, pos_tag
# Download required resources (including the missing punkt_tab)
nltk.download('punkt')
nltk.download('punkt_tab')  # <--- Added this line to fix the LookupError
nltk.download('averaged_perceptron_tagger')
nltk.download('averaged_perceptron_tagger_eng')
# Accept user input
sentence = input("Enter a sentence: ")
# Step 1: Tokenization
tokens = word_tokenize(sentence)
print("\nTokens:", tokens)
# Step 2: PoS Tagging                                #11249A290
tagged = pos_tag(tokens)
# Display Output
print("\nWord\t\tPoS Tag")
print("-" * 30)
for word, tag in tagged:
    print(f"{word}\t\t{tag}")
```

```python
import nltk
from nltk import word_tokenize
from nltk.tag import RegexpTagger
nltk.download('punkt')                #11249A312
nltk.download('punkt_tab')
patterns = [
    (r'.*ing$', 'VBG'),              # gerunds -> running
    (r'.*ed$', 'VBD'),               # past tense -> jumped
    (r'.*es$', 'VBZ'),               # 3rd person singular -> goes
    (r'.*ould$', 'MD'),              # modals -> would, could
    (r'.*\'s$', 'NN$'),              # possessive nouns
    (r'.*s$', 'NNS'),                # plural nouns
    (r'^-?[0-9]+(\.[0-9]+)?$', 'CD'), # cardinal numbers
    (r'(The|the|A|a|An|an)$', 'DT'),  # determiners
    (r'(in|on|at|of|for|with)$', 'IN'),# prepositions
    (r'.*ly$', 'RB'),                # adverbs
    (r'.*', 'NN'),                   # default: noun
    ]
rule_based_tagger = RegexpTagger(patterns)
sentence = input("Enter a sentence: ")
tokens = word_tokenize(sentence)
tagged_output = rule_based_tagger.tag(tokens)
print("\nWord\t\tPoS Tag (Rule-Based)")
print("-" * 35)
for word, tag in tagged_output:
    print(f"{word}\t\t{tag}")
```

```python

states = ['DT', 'NN', 'VB']
start_p = {'DT': 0.6, 'NN': 0.3, 'VB': 0.1}
trans_p = {
    'DT': {'DT': 0.05, 'NN': 0.85, 'VB': 0.10},
    'NN': {'DT': 0.10, 'NN': 0.30, 'VB': 0.60},
    'VB': {'DT': 0.40, 'NN': 0.50, 'VB': 0.10},
}
emit_p = {
    'DT': {'the': 0.7, 'a': 0.3},
    'NN': {'dog': 0.4, 'cat': 0.4, 'bites': 0.05, 'runs': 0.15},
    'VB': {'bites': 0.6, 'runs': 0.4},
}
def viterbi(obs, states, start_p, trans_p, emit_p):
    V = [{}]
    path = {}

    for st in states:
        V[0][st] = start_p[st] * emit_p[st].get(obs[0], 1e-6)
        path[st] = [st]

    for t in range(1, len(obs)):
        V.append({})
        new_path = {}
        for cur in states:
            best_prob, best_prev = max(
                (V[t-1][prev] * trans_p[prev][cur] * emit_p[cur].get(obs[t], 1e-6),
prev)
                for prev in states)
            V[t][cur] = best_prob
            new_path[cur] = path[best_prev] + [cur]
        path = new_path

    n = len(obs) - 1
    best_final_prob, best_final_st = max((V[n][st], st) for st in states)
    return best_final_prob, path[best_final_st]
sentence = ['the', 'dog', 'bites']
prob, tag_sequence = viterbi(sentence, states, start_p, trans_p, emit_p)
print("Sentence :", sentence)
print("Best Tag Sequence :", tag_sequence)
print(f"Probability : {prob:.6f}")
```

```python

import nltk
from nltk.corpus import brown
from nltk import UnigramTagger, BigramTagger, word_tokenize   #11249A312

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')

category = 'news'
tagged_sentences = brown.tagged_sents(categories=category)
size_90 = int(len(tagged_sentences) * 0.9)
train_data_90 = tagged_sentences[:size_90]
test_data_90 = tagged_sentences[size_90:]

unigram_tagger_90 = UnigramTagger(train_data_90)
bigram_tagger_90 = BigramTagger(train_data_90, backoff=unigram_tagger_90)
print(f"--- Results for '{category}' category with 90% Train Split ---")
print('Unigram accuracy:', unigram_tagger_90.accuracy(test_data_90))
print('Bigram accuracy :', bigram_tagger_90.accuracy(test_data_90))

size_50 = int(len(tagged_sentences) * 0.5)
train_data_50 = tagged_sentences[:size_50]
test_data_50 = tagged_sentences[size_50:]
unigram_tagger_50 = UnigramTagger(train_data_50)
bigram_tagger_50 = BigramTagger(train_data_50, backoff=unigram_tagger_50)
print(f"\n--- Results for '{category}' category with 50% Train Split ---")
print('Unigram accuracy:', unigram_tagger_50.accuracy(test_data_50))
print('Bigram accuracy :', bigram_tagger_50.accuracy(test_data_50))

alt_category = 'romance'
alt_sentences = brown.tagged_sents(categories=alt_category)
alt_size = int(len(alt_sentences) * 0.9)
alt_train = alt_sentences[:alt_size]
alt_test = alt_sentences[alt_size:]
alt_unigram = UnigramTagger(alt_train)
alt_bigram = BigramTagger(alt_train, backoff=alt_unigram)
print(f"\n--- Cross-Domain Results for '{alt_category}' category with 90% Train Split ---")
print('Unigram accuracy:', alt_unigram.accuracy(alt_test))
print('Bigram accuracy :', alt_bigram.accuracy(alt_test))
```

```python

import nltk
from nltk.corpus import brown                    #11249A312
from nltk import UnigramTagger, word_tokenize
from nltk.tag import brill, brill_trainer

nltk.download('brown')
nltk.download('punkt')
nltk.download('punkt_tab')

tagged_sentences = brown.tagged_sents(categories='news')
size = int(len(tagged_sentences) * 0.9)
train_data = tagged_sentences[:size]
test_data = tagged_sentences[size:]

baseline_tagger = UnigramTagger(train_data)
print(f"Baseline (Unigram) Accuracy: {baseline_tagger.accuracy(test_data)*100:.2f}%")

templates = brill.brill24()

print("\n--- Training Brill Tagger (max_rules = 10) ---")
trainer_10 = brill_trainer.BrillTaggerTrainer(baseline_tagger, templates, trace=3)
brill_tagger_10 = trainer_10.train(train_data, max_rules=10)
print(f"Brill Tagger Accuracy (10 rules): {brill_tagger_10.accuracy(test_data)*100:.2f}%")

print("\n--- Training Brill Tagger (max_rules = 30) ---")
trainer_30 = brill_trainer.BrillTaggerTrainer(baseline_tagger, templates, trace=0)
brill_tagger_30 = trainer_30.train(train_data, max_rules=30)
print(f"Brill Tagger Accuracy (30 rules): {brill_tagger_30.accuracy(test_data)*100:.2f}%")

test_sentence = "They will fish for hours near the river bank"
tokens = word_tokenize(test_sentence)
print("\nTagged Output (30 rules):")
print(brill_tagger_30.tag(tokens))
```
