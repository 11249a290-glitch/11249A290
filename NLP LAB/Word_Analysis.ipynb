```python
import nltk
import spacy
from nltk.stem import WordNetLemmatizer                                                             #11249a290
from nltk.tokenize import word_tokenize
nltk.download("punkt", quiet=True)
nltk.download("punkt_tab", quiet=True)
nltk.download("wordnet", quiet=True)
nlp = spacy.load("en_core_web_sm")
wordnet_lemmatizer = WordNetLemmatizer()
text = "The children went to the library because they saw many geese flying over the trees."
print(f"Original Text:\n{text}\n")
print(f"{'Token':15} | {'spaCy Lemma':15} | {'NLTK WordNet (Default)':25}")
print("-" * 62)
spacy_doc = nlp(text)
spacy_lemmas = {token.text: token.lemma_ for token in spacy_doc}
nltk_tokens = word_tokenize(text)
for token in nltk_tokens:
    if token in [".", ",", "!", "?"]:
        continue
    spacy_res = spacy_lemmas.get(token, token.lower())
    nltk_res = wordnet_lemmatizer.lemmatize(token.lower())
    match_status = "" if spacy_res == nltk_res else "❌ Mismatch"
    print(f"{token:15} | {spacy_res:15} | {nltk_res:25} {match_status}")
```

```python
class LexiconFST:
    def __init__(self):
        self.lexicon = [
            ("go", "go", "V+BASE"),                                                                           //11249a290
            ("went", "go", "V+PAST"),
            ("going", "go", "V+PROG"),
            ("eat", "eat", "V+BASE"),
            ("ate", "eat", "V+PAST"),
            ("eating", "eat", "V+PROG"),
            ("walk", "walk", "V+BASE"),
            ("walked", "walk", "V+PAST"),
            ("walking", "walk", "V+PROG"),
            ("run", "run", "V+BASE"),
            ("ran", "run", "V+PAST"),
            ("running", "run", "V+PROG"),
            ("see", "see", "V+BASE"),
            ("saw", "see", "V+PAST"),
            ("seeing", "see", "V+PROG"),
            ("speak", "speak", "V+BASE"),
            ("spoke", "speak", "V+PAST"),
            ("speaking", "speak", "V+PROG"),
            ("write", "write", "V+BASE"),
            ("wrote", "write", "V+PAST"),
            ("writing", "write", "V+PROG"),
            ("take", "take", "V+BASE"),
            ("took", "take", "V+PAST"),
            ("taking", "take", "V+PROG"),
            ("mouse", "mouse", "N+SG"),
            ("mice", "mouse", "N+PL"),
            ("goose", "goose", "N+SG"),
            ("geese", "goose", "N+PL"),
            ("child", "child", "N+SG"),
            ("children", "child", "N+PL"),
            ("ox", "ox", "N+SG"),
            ("oxen", "ox", "N+PL"),
            ("cat", "cat", "N+SG"),
            ("cats", "cat", "N+PL"),
            ("dog", "dog", "N+SG"),
            ("dogs", "dog", "N+PL"),
            ("fox", "fox", "N+SG"),
            ("foxes", "fox", "N+PL")
        ]
        self.analysis_table = {}
        self.generation_table = {}
        self._compile_fst()
    def _compile_fst(self):
        for surface, lemma, tag in self.lexicon:
            self.analysis_table[surface.lower()] = f"{lemma}{tag}"
            analysis_key = f"{lemma.lower()}{tag}"
            self.generation_table[analysis_key] = surface
    def analyze(self, surface_word):
        return self.analysis_table.get(surface_word.lower(), "NOT_FOUND")
    def generate(self, lexical_string):
        return self.generation_table.get(lexical_string.lower(), "NOT_FOUND")
fst = LexiconFST()
print("=== 1. MORPHOLOGICAL ANALYSIS DEMO ===")
analysis_tests = ["mice", "went", "cats", "running", "children", "spoke", "foxes", "unknown"]
for word in analysis_tests:
    print(f"Surface: {word:10} ➔ Analysis: {fst.analyze(word)}")
print("\n=== 2. MORPHOLOGICAL GENERATION DEMO ===")
generation_tests = ["mouseN+PL", "goV+PAST", "catN+PL", "runV+PROG", "childN+PL", "speakV+PAST", "foxN+PL", "dogN+SG"]
for lex in generation_tests:
    print(f"Lexical Input: {lex:15} ➔ Surface Word: {fst.generate(lex)}")
```

```python
import nltk
from nltk.stem import PorterStemmer, SnowballStemmer                             #11249a290
nltk.download('punkt', quiet=True)
p_stemmer = PorterStemmer()
s_stemmer = SnowballStemmer("english")
test_words = [
    "generously",
    "generosity",
    "knives",
    "fairly",
    "singing",
    "singed",
    "matrices",
    "matrix",
    "provisional",
    "community"
]
print(f"{'Original Word':15} | {'Porter Stem':15} | {'Snowball Stem':15} | {'Match?'}")
print('-' * 65)
for w in test_words:
    p_out = p_stemmer.stem(w)
    s_out = s_stemmer.stem(w)
    match_status = "Yes" if p_out == s_out else "No ❌"
    print(f"{w:15} | {p_out:15} | {s_out:15} | {match_status}")
```
